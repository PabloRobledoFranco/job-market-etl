```python
df_skills = pd.read_csv(skills_path)

print("CSV shape: ", df_skills.shape)
print("\nNulls:\n", df_skills.isnull().sum())
print("\nUnique job_link: ", df_skills["job_link"].nunique())
df_skills.head(3)
```
